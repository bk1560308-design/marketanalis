```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
```

```python
le_state = LabelEncoder()
le_city = LabelEncoder()
le_type = LabelEncoder()
```

```python
data['State'] = le_state.fit_transform(data['State'])
data['City'] = le_city.fit_transform(data['City'])
data['Startup_Type'] = le_type.fit_transform(data['Startup_Type'])
```

```python
import pickle

# 🔹 Save encoders
pickle.dump(le_state, open('le_state.pkl', 'wb'))
pickle.dump(le_city, open('le_city.pkl', 'wb'))
pickle.dump(le_type, open('le_type.pkl', 'wb'))

# 🔹 Save unique values (for UI)
unique_values = {
    "State": list(le_state.classes_),
    "City": list(le_city.classes_),
    "Startup_Type": list(le_type.classes_)
}

pickle.dump(unique_values, open('unique_values.pkl', 'wb'))
```

```python
data
```

```python
x=data.drop('Success',axis=1)
x
```

```python
y=data['Success']
y
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.8,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
x_test
```

```python
x_train
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
```

```python
model_random=RandomForestClassifier()
model_random.fit(x_train,y_train)
```

```python
model_logistic=LogisticRegression(max_iter=1000)
model_logistic.fit(x_train,y_train)
```

```python
import pickle

pickle.dump(model_logistic, open('model.pkl', 'wb'))
pickle.dump(scaler, open('scaler.pkl', 'wb'))
```

```python
pred_r = model_random.predict(x_test)
```

```python
pred_l = model_logistic.predict(x_test)
```

```python
y_prob = model_logistic.predict_proba(x_test)
print(y_prob)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
print("FOR RANDOM FOREST")
print("Accuracy:", accuracy_score(y_test, pred_r))
print("Confusion Matrix:\n", confusion_matrix(y_test, pred_r))
print("Report:\n", classification_report(y_test, pred_r))
```

```python
print("FOR LOGISTIC REGRESSION")
print("Accuracy:", accuracy_score(y_test,pred_l))
print("Confusion Matrix:\n", confusion_matrix(y_test,pred_l))
print("Report:\n", classification_report(y_test,pred_l))
```
