```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x='Success', y='Investment_Amount', data=startup)

plt.title("Investment vs Success")
plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.barplot(x='Success', y='Investment_Amount', data=startup)

plt.title("Average Investment by Success")
plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(x='Success', data=startup)

plt.title("Startup Success Count")
plt.show()
```
