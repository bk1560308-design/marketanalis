```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(x='Average_Income', y='Average_Spending', data=population)

plt.title("Income vs Spending")
plt.show()
```

```python
top_states = population.groupby('State')['Average_Spending'].mean().nlargest(10).reset_index()

plt.figure(figsize=(10,5))

sns.barplot(x='State', y='Average_Spending', data=top_states)

plt.title("Top States by Spending")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(population['Average_Spending'], bins=30, kde=True)

plt.title("Spending Score Distribution")
plt.show()
```
